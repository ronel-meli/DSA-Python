```python
days = int(input("Enter no. of days: "))
seconds = days * 24 * 60 * 60
print("Number of second:", seconds)




```

```python
import math

radius = float(input("Enter radius: "))
volume = (4/3) * math.pi * radius**2
print("Volume:", volume)
```

```python
def area(side):
    return side * side

def perimeter(side):
    return 4 * side

side = float(input("Enter side length: "))
print("Area:", area(side))
print("Perimeter:", perimeter(side))
```

```python
def check_case(char):
    if char.isupper():
        print("The character is uppercase")
    elseif char.islower():
        print("The character is lowercase")
    else:
        print("That is not a letter")

char = input("Enter a character: ")
check_case(char)
```

```python
x = 0
y = 20

while not (y < 6):
    y = y - 4
    x = x + (2 / y)

print("x =", x)
```

```python
numbers = []

for i in range(5):
    value = float(input("Enter value " + str(i+1) + ": "))
    numbers.append(value)

average = sum(numbers) / len(numbers)

print("Values entered:", numbers)
print("Average:", average)
```
